測試一：使用原始資料 (4 人，折價券 100 元)

```python
# 1. 原始輸入資料
event_name = "  WEEKEND DINNER  "
meal_amount_str = " NT$1,200 "
people_str = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 2. 資料清理與型別轉換
# 去除 event_name 頭尾空白並轉成小寫
cleaned_event_name = event_name.strip().lower()

# 移除 meal_amount 中的 "NT$"、逗號與空白，並轉成浮點數 (float)
cleaned_meal_str = meal_amount_str.replace("NT$", "").replace(",", "").strip()
meal_amount = float(cleaned_meal_str)

# 將文字人數轉成整數 (int)
people = int(people_str)

# 3. 分步計算費用
service_fee = meal_amount * service_rate             # 服務費：原始餐費 × 服務費率
total_amount = meal_amount + service_fee - discount   # 應付總額：原始餐費 ＋ 服務費 － 折價券
per_person_fee = total_amount / people                # 每人費用：應付總額 ÷ 分攤人數

# 4. 判斷是否超過預算 (布林變數)
is_over_budget = per_person_fee > budget_per_person

# 5. 輸出費用摘要 (使用 f-string 格式化到小數點後兩位)
print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people} 人")
print(f"每人費用：{per_person_fee:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
